```python
from pydantic import BaseModel
from openai import OpenAI
from datetime import date
```

```python
client = OpenAI()
```

```python
class InvoiceLine(BaseModel):
    description: str
    quantity: float
    unit_price: float
    amount: float

class Invoice(BaseModel):
    vendor : str
    total : float
    invoice_number : str
    currenct: str
    due_date: date | None
    invoice_lines : list[InvoiceLine]
```

```python
input_text = """Hey, got the latest bill from Acme Industrial.
Invoice is 84729-A. Total came out to $4,821.17.
It's due the 15th of next month. They are billing me for 
two fidget spinners. They are $10 each."""
```

```python
response = client.responses.parse(
    model='gpt-5.6-sol',
    input = input_text,
    text_format = Invoice
)
```

```python
invoice = response.output_parsed

print(invoice.vendor)
print(invoice.invoice_number)
print(invoice.total)
print(invoice.invoice_lines)
print(invoice.due_date)
```
